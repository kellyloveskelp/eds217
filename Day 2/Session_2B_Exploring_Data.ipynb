```python
import pandas as pd

url = 'https://eds-217-essential-python.github.io/data/openaq_goleta_measurments.csv'
goleta = pd.read_csv(url)
```

```python
goleta.tail() # .tail() does the opposite of .head() ! It calls the last number of rows
```

```python
# .shape gives you (rows, columns) 
goleta.shape # .shape is a **tupple**
# no "()" because position matters 
# and only returns 2 arguements
```

```python
# Remember f strings
print(f"(rows, columns) = {goleta.shape}")
```

```python
# .shape and .dtype is an ATTRIBUTE
goleta.dtypes
```

```python
# .info() is a METHOD
# combines the shape, the column names, the types,
# AND how many non-null values each column has
goleta.info()
```

```python
goleta.isnull().sum() # asks the same question more directly
```

```python
url_sb = 'https://eds-217-essential-python.github.io/data/openaq_santa_barbara_measurments.csv'
santa_barbara = pd.read_csv(url_sb)
santa_barbara.info()
# same three columns are empty: country_iso, isMobile, isMonitor
```

```python
# To look up an object, call "?" first
# ?goleta.info
# can call a method definition, but must be an attribute, aka no "()"
```

```python
# .describe() gives you count, mean, standard deviation, min, max, 
# and quartiles for every numeric column
goleta.describe()

# .describe() averages whatever numbers are in a column, 
# with no check on whether they belong together 
# therefore, grouping/mutating may be neccessary before calling .describe()
```

```python
# .describe() is for numbers
# For text columns, the question is usually “what values appear, and how often?”
# .value_counts() is the answer
goleta['parameter'].value_counts()
# shows that parameter contains three different pollutants: pm25, o3, and pm10
```

```python
## Renaming column with a dictionary
# A dictionary stores pairs. 
# Each pair has a key and a value, written "key:value" 
# and the whole thing goes in curly braces
new_names = {
    'datetimeUtc': 'datetime_utc',
    'datetimeLocal': 'datetime_local',
}
new_names
# A dictionary is a set of labelled pairs, and you call indv objects out of it by name
```

```python
# This is the easiest way to rename columns
goleta_renamed = goleta.rename(columns=new_names)
goleta_renamed.columns.tolist()
```

```python
goleta.rename(columns={'value': 'concentration'}).columns.tolist()
```

```python
goleta_renamed = goleta.rename(columns={
    'value': 'concentration',
    'parameter' : 'pollutant'
    }).columns.tolist()
goleta_renamed
```
