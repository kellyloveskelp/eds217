```python
#Use .value_counts() on LTER_Site and on Station. What does each column tell you?
print(toolkit['LTER_Site'].value_counts())
print(toolkit['Station'].value_counts())
# Tells us that there are values in every row of the column
```

```python
# Use .value_counts() on Flag_Daily_AirTemp_Mean_C. What values appear, and what do you think they mean?
toolkit['Flag_Daily_AirTemp_Mean_C'].value_counts()
# Values labeled "E" appear
# Could indicate "Error"
```

```python
# Use .value_counts() on the Year column and look at both ends of the result, with .head() and then with .tail(). 
# Do all years have the same number of observations? What would explain a year with fewer?
toolkit['Year'].value_counts().tail()

# Years with 366 samples were likely leap-years
# Year with only 214 samples(1988) could've been the year the study was started
```

```python
# What is the mean daily air temperature across the whole record? The minimum? The maximum?
print("Mean = ", toolkit['Daily_AirTemp_Mean_C'].mean())
print("Minimum = ", toolkit['Daily_AirTemp_Mean_C'].min())
print("Maximum = ", toolkit['Daily_AirTemp_Mean_C'].max())
```

```python
# Are the mean, the minimum and the maximum plausible for a site above the Arctic Circle? Say why or why not.
print("Median = ", toolkit['Daily_AirTemp_Mean_C'].median())
toolkit['Daily_AirTemp_Mean_C']

# The mean, min, and max appear plausible based off of the median and number of times Daily_AirTemp_Mean_C was reported to be less then -40 degC.
```
