```python
import pandas as pd
```

```python
url = 'https://eds-217-essential-python.github.io/data/openaq_goleta_measurments.csv'
goleta = pd.read_csv(url)
goleta.head(4) # can insert a number into head() and it will return that number of rows
# indexes starting at 0, i.e. head(4) outputs 0:3
```

```python
url_sb = 'https://eds-217-essential-python.github.io/data/openaq_santa_barbara_measurments.csv'
santa_barbara = pd.read_csv(url_sb)
santa_barbara.head(4) # can insert a number into head() and it will return that number of rows
# indexes starting at 0, i.e. head(4) outputs 0:3
```

```python
goleta['value'] # pulling one **series** out of the dataframe
goleta['value'].head()
```

```python
print(type(goleta)) # Dataframe
print(type(goleta['value'])) # Series
```

```python
# Square brackets with one column name give you a Series. 
# Square brackets with a list of column names give you back a DataFrame, 
# even if the list has one item:

goleta['value']       # Returns a Series
goleta[['value']]     # Returns a DataFrame with one column

# OR, define the liste seperately
list_of_columns = ['value']
dataframe = goleta[list_of_columns]
type(dataframe)
```

```python
lat = goleta['latitude']       # Returns a Series
lat_lon = goleta[['latitude', 'longitude']]  
print(type(lat))
print(type(lat_lon))
```

```python
# .columns attribute gives you the names
goleta.columns
```

```python
# The result is a pandas object rather than a plain Python list. 
# To get the list, which is easier to read and easier to work with, use .tolist()
column_names = goleta.columns.tolist()
column_names

# if you ever want to turn an object type into a different type ".to_Ex" usually does the trick
# ** .to_excel ** turns dataframes into excel spreadsheets
```

```python
# You can also get items out by position, counting from zero:
print(column_names[0])
print(column_names[1])
```

```python
# The length function "len()" tells you how many ROWS there are in a dataframe
len(goleta)
```

```python
keep = ['location_name', 'parameter', 'value', 'unit'] # list of columns to keep
goleta[keep] # new dataframe with only listed columns
```
