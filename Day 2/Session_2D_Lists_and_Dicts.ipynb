# 1. Making lists

```python
names = ["Pricilla", "Emalia", "Kailani", "Aaron", "Natasha", "K-Loo"]
```

# 2. Getting things out of a list

```python
names.sort()
names
```

# 3. Changing a list

```python
names[0]= "Liam"
names
```

```python
new_students = ["Nina", "Ama"]
```

```python
names = names + new_students
names
```

```python
newer_students = ["Elyse", "Karolina"]
names[:2] = newer_students
names
```

# 4. Making a list

```python
for name in names:
    print(name)
```

```python
for name in names:
    print(f"{name} is a great student")
```

# 5. Making a dictionary

```python
name_pets = {'Nina': 1, 'Precilla': 7, 'Elyse': 5, 'Karolina': 0}
```

```python
name_pets['Karla'] = 1
```

```python
name_pets
```

```python
# With dictionaries, you have to call the key rather than the index
name_pets['Elyse']
```

# 6. Getting things out of a dictionary

```python
for key, value in name_pets.items():
    print(f"{key} has {value} pets")
    if value > 1:
        print(f"{key} has too many pets!")
```
