```python
import pandas as pd

base = 'https://eds-217-essential-python.github.io/data/'

goleta = pd.read_csv(base + 'openaq_goleta_measurments.csv')
santa_barbara = pd.read_csv(base + 'openaq_santa_barbara_measurments.csv')
cnsi = pd.read_csv(base + 'openaq_CNSI_measurments.csv')

print(goleta.shape)
print(santa_barbara.shape)
print(cnsi.shape)
```

```python
## pd.merge() CANNOT stack tables. 
# pd.concat() CAN, and it takes a *list* of DataFrames:
aq = pd.concat([goleta, santa_barbara, cnsi])

aq.shape
```

```python
## Label your pieces before you stack them so you don't loose track of the origin
goleta['site'] = 'Goleta'
santa_barbara['site'] = 'Santa Barbara'
cnsi['site'] = 'CNSI'

aq = pd.concat([goleta, santa_barbara, cnsi])
aq.shape
```

```python
## Let's double check our numbers
print("SB: ", santa_barbara.shape)
print("Goleta: ", goleta.shape)
print("CSNI: ", cnsi.shape)
print(aq['site'].value_counts())
```

```python
## pd.concat() by default preserves the index values of each dataframe...
# This means indices can repeat!
# Especially if indicies are row numbers
aq.loc[0]
# three zero locations
```

### write the stacking pattern, pd.concat([a, b, c]), and say what ignore_index=True fixes


```python
# Use a new index to create a new index when using pd.concat()
# Set ignore_index=True
aq = pd.concat([goleta, santa_barbara, cnsi], ignore_index=True)
aq.loc[0]
# only one zero location
```

#### ignore_index=True fixes the index order- don't forget!

### describe the difference between a long table and a wide one, and say what each is good at


A **long** table can be appended to forever without ever changing its columns, BUT a long table is awkward to compare across. To make this easier a **wide** table allows you to make a comparison by reading two numbers that are already sitting next to each other.

```python
aq[['site', 'parameter', 'value', 'unit', 'datetimeLocal']].head()
```

### write the pivot pattern, pivot_table(index=, columns=, values=), and read its result


```python
## pivot_table() rotates a DataFrame from long to wide

# df.pivot_table(index='rows', columns='columns', values='numbers')
# #              ↑             ↑                  ↑
# #              what labels   what labels        what goes
# #              the rows      the columns        in the cells
```

```python
aq.pivot_table(
    index='site',
    columns='parameter',
    values='value'
)
## **by default, pivot_table() takes the average of the column values**
```

```python
## use agg.func= to assign a different statistical method
aq_count = aq.pivot_table(
    index='site', columns='parameter', values='value', aggfunc='count')
aq_count
```

```python
means = aq.pivot_table(index='site', columns='parameter', values='value')
means
# site becomes index
```

```python
# use .reset.index() to reset the site to a column
flat = means.reset_index()
flat
```

```python
## Practice:
swap_aq = aq.pivot_table(
    index='parameter',
    columns='site',
    values='value',
    aggfunc='mean'
)
swap_aq
```


turn a .value_counts() result back into a table you can merge with, using .reset_index()


```python
# pd.merge() needs two DataFrames with columns to join on 
# and a Series has an index where the column would be. 
# So we promote it into a DataFrame:

counts = aq['site'].value_counts().reset_index()
counts
```

```python
counts = counts.rename(columns={'count': 'n_readings'})
counts
```

### Key points
pd.concat([a, b, c]) stacks tables with the same columns, top to bottom. It matches nothing and drops nothing.

Label the pieces before you stack them. Once the rows are combined, there is no way to recover which file a row came from.

Use ignore_index=True so the stacked table gets one clean set of row numbers.

A long table has one measurement per row and its labels in columns. It is easy to append to and hard to compare across.

A wide table has one row per thing and one column per measurement. It is easy to compare and awkward to extend.

The pivot pattern is df.pivot_table(index=, columns=, values=). It goes long to wide.

pivot_table averages by default. Pass aggfunc='count' and look at that version first.

The result is indexed by the index= column. .reset_index() turns it back into a plain table.

.value_counts().reset_index() promotes a count Series into a two-column table you can merge with. Rename the count column while you are there.
