### Clean it !


```python
data.shape
```

```python
data_dedup = data.drop_duplicates() 
data_dedup
## 300 rows|
```

```python
data_dedup['site'] = data_dedup['site'].str.lower()  
data_dedup['site'] = data_dedup['site'].str.strip() 
data_dedup['site'] = data_dedup['site'].str.replace('site-', 'site_')
data_dedup['site'].value_counts()
```

```python
data_dedup['pH'] = data_dedup['pH'].str.replace(',', '.').astype(float)
data_dedup.describe()
```

```python
cols = ['temperature_c', 'dissolved_oxygen_mg_L', 'conductivity_uS_cm']
data_dedup = data_dedup.dropna(subset=cols)
data_dedup.shape
#36
```

```python
data_dedup['n_replicates'] = data_dedup['n_replicates'].fillna(1).astype(int)
data_dedup['n_replicates'].value_counts(ascending=False)
```

```python
# df_filtered = df[df['column'] > x]
data_filtered = data_dedup[data_dedup['temperature_c'] > -100].copy()
data_filtered
#9
```

```python
data_filtered['temperature_c'].describe()
```

The new mean is 19.22, whereas the old mean was -11.21.

```python
data_rename = data_filtered.rename(columns={'collection date': 'collection_date'})
data_rename.head(1)
```

```python
data_rename['conductivity_mS_cm'] = data_rename['conductivity_uS_cm'] / 1000
```

```python
data_rename['temperature_f'] = (data_rename['temperature_c'] * 9 / 5) + 32
data_rename.head()
```

```python
def celsius_to_fahrenheit(celsius):
    fahrenheit = (celsius * 9 / 5) + 32
    return fahrenheit

data_rename['temperature_f'] = data_rename['temperature_c'].apply(celsius_to_fahrenheit)
data_rename.head()
```

```python
def classify_ph(pH):
    if pH < 6.5:
        return 'acidic'
    elif pH > 7.5:
        return 'alkaline'
    else:
        return 'neutral'

data_rename['ph_class'] = data_rename['pH'].apply(classify_ph)
data_rename['ph_class'].value_counts()
```

```python
data_acidic = data_rename[data_rename['ph_class'] == "acidic"].copy()
data_acidic['site'].value_counts()
```

```python
site_d = data_rename[data_rename['site'] == "site_d"]
site_f = data_rename[data_rename['site'] == "site_f"]
print("Site D Mean= ", site_d['dissolved_oxygen_mg_L'].mean())
print("Site F Mean= ", site_f['dissolved_oxygen_mg_L'].mean(), "#low")
```
