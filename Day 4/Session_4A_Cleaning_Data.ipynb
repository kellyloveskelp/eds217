```python
import pandas as pd

url = 'https://eds-217-essential-python.github.io/data/marine_microplastics.csv'
plastics = pd.read_csv(url)
plastics.head()
```

```python
sea = plastics[plastics['Organization'] == "Sea Education Association"]
sea
```

```python
plastics.columns.tolist()
```

```python
plastics.isnull().sum()
#  .isnull() returns a True/False for every cell, 
# and .sum() counts the Trues, one column at a time
```

```python
plastics.isnull().sum().sort_values(ascending=False).head(5)
```

```python
## .dropna()
plastics.dropna().shape
# Bare .dropna() is almost always the wrong tool
# With no arguments, .dropna() removes every row that has a missing value in any column at all
```

```python
# subset= takes a list of column names, 
# and only a gap in one of those columns can delete a row:
measured = plastics.dropna(subset=['Measurement'])
measured.shape
```

```python
plastics['Unit'].value_counts()
```

```python
measured['Unit'].value_counts()
```

```python
## Filling gaps with .fillna()
# .fillna() replaces the gaps with a value you choose
plastics['Regions'] = plastics['Regions'].fillna('Unspecified')
plastics['Regions'].value_counts().head(5)

## Kelly Caylor ""
# filling any dataset with new data should be doe with the upmost consideration and care
# dropna and fillna and are "powerful" methods 
# that should only be used when you know EXACTLY what youre doing
```

```python
plastics.isnull().sum().sort_values(ascending=False).head(3)
```

```python
# The other classic defect is the same sample entered twice
# .duplicated() returns a mask, one True or False per row, and .sum() counts them
plastics.duplicated().sum()
```

```python
plastics[['OBJECTID', 'GlobalID', 'Latitude', 'Longitude', 'Date', 'Measurement']].head(3)
# OBJECTID and GlobalID are identifiers, assigned one per record when the archive was built. 
# No two rows can ever share them, so no two rows can ever be identical, 
# so .duplicated() can only ever return zero.
# Any table with an auto-generated ID column will report zero duplicates forever.
```

```python
# need to use "subset=" to point duplicate towards the right columns
plastics.duplicated(subset=sample_columns).sum()
# .duplicated() is only as good as the columns you point it at
```

### .duplicated() (and realted methods) can be used to create a filter that you then apply to the dataset

```python
# Define desired columns 
sample_columns = ['Latitude', 'Longitude', 'Date', 'Measurement', 'Unit']

# Create a filter (True/False) using these columns
my_filter = plastics.duplicated(subset=sample_columns, keep=False)

plastics[my_filter]
```

```python
# .drop_duplicates() takes the same argument and returns a table with the repeats removed, 
# keeping the first of each
deduplicated = plastics.drop_duplicates(subset=sample_columns)
deduplicated.shape
```

```python
repeats = plastics[plastics.duplicated(subset=sample_columns, keep=False)]
print(repeats.shape)
repeats[['Measurement', 'Unit', 'Sampling Method', 'Short Reference']].head(6)
```

```python
(repeats['Measurement'] == 0).sum()
```

```python
# dtypes versus astypes

print(plastics.dtypes)
plastics['Accession Number'].head(3)
# showing up as int64 but should be string
```

```python
# .astype() converts a column to the type you name
plastics['Accession Number'] = plastics['Accession Number'].astype(str)
# turns "Accession Number" into a string
plastics['Accession Number'].head(3)
```

```python
# Easiest object types: .astype(float), .astype(int), and .astype(str)
```

### Key points

Run df.isnull().sum() first, every time, on any table you did not build yourself.

Bare .dropna() deletes a row for a gap in any column. In this file that cost 96% of the data. Use .dropna(subset=['col']) and name the columns you actually need.

Before you drop, count what you are dropping and look at what it has in common. Missingness usually lines up with something.

.fillna(value) replaces gaps with a value you choose. Choose one that cannot be mistaken for data.


.duplicated() and .drop_duplicates() both take subset=[...]. Without it, any table with an ID column reports zero duplicates forever.

.astype(str/int/float) converts a column. It returns a new column, so assign it back, and remember that narrowing a type throws information away.
