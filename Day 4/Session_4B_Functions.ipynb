### A **function** is the cure to the tedium and risk of a copy-and-paste lifestyle. You write the work once, give it a name, and then you have a new tool that you and pandas can both use.

```python
import pandas as pd

url = 'https://eds-217-essential-python.github.io/data/marine_microplastics.csv'
plastics = pd.read_csv(url)
plastics.shape
```


#### define a function with def and hand a value back with return


```python
# def and return
def celsius_to_fahrenheit(celsius):
    fahrenheit = (celsius * 9 / 5) + 32
    return fahrenheit

# def announces that a definition is coming.
# celsius_to_fahrenheit is the name. You choose it, and you should choose it well.
# (celsius) is the parameter: a name for the value that will be handed in.
# everything indented under the def line is the body, and return says what comes back.

# A function without return still runs, but it hands back None. 
# If your function seems to “do nothing,” a missing return is the first thing to check.
```

```python
celsius_to_fahrenheit(0)
```


#### call your function, and store what it gives you


```python
# meters to feet conversion
def meters_to_feet(meters):
    feet = (meters * 3.28084)
    return feet
meters_to_feet(1000)
```

```python
# When writing functions, use descriptive language
def ppm_to_ppb(concentration_ppm):
    """Convert a concentration from parts per million to parts per billion."""
    return concentration_ppm * 1000

    # The line in triple quotes is a docstring.
    # Goes in the first line of the body and will pop up when you call ? on a function

# ?ppm_to_ppb
```

#### call a function with keyword arguments, so the reader can see what each one means


```python
# def function(parameters, keyword= arguement):
def convert_depth(depth, units='meters'):
    """Convert a depth in metres to metres, feet, or fathoms."""
    if units == 'meters':
        return depth
    elif units == 'feet':
        return depth * 3.28084
    elif units == 'fathoms':
        return depth * 0.546807
    else:
        return None

# every function has some sort of positional parameters and optional keywords
# keyword arguements override the order you write them in
```

```python
plastics['Latitude'].head()
```

```python
def classify_latitude(latitude):
    """Label a latitude as tropical, temperate, or polar."""
    if -23.5 <= latitude <= 23.5:
        return 'tropical'
    elif latitude < -66.5 or latitude > 66.5:
        return 'polar'
    else:
        return 'temperate'
classify_latitude(50)
```


#### run a function down an entire column with .apply()

```python
# .apply() runs a function once for every value in a column 
# and gives you back a column of the results
plastics['Latitude'].apply(classify_latitude).head()
```

```python
# .apply() follows the same rule as almost everything else in pandas: 
# it hands back a new column and leaves the original table alone. 
# Assign it to keep it:

plastics['zone'] = plastics['Latitude'].apply(classify_latitude)
plastics['zone'].value_counts()
```

### Key points
def name(parameter): opens a function; the indented body is what it does; return is what it hands back. No return means it hands back None.

Defining a function runs nothing. Calling it does.

Name functions for what they produce. Add a docstring when the name is not enough.

Parameters can have defaults, and every defaulted parameter comes after the undefaulted ones.

Keyword arguments name the value at the call site. You have been using other people’s keyword arguments all week.

df['col'].apply(function_name) runs your function down a whole column. Pass the name, with no parentheses of its own.

Write a function the second time you need something, not the fifth.
