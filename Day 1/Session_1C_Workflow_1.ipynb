```python
## STEP 1:  IMPORT
import pandas as pd
import matplotlib.pyplot as plt

url = "https://eds-217-essential-python.github.io/data/toolik_weather.csv"
df = pd.read_csv(url)
```

```python
## STEP 2: EXPLORE
df.head()
```

```python
df.describe()
# Provides summary statistics for the dataset
```

```python
## STEP 3: CLEAN

# The "health check"
df.isnull().sum()
# Returns the count of empty values
```

```python
## STEP 7: GROUP
monthly = df.groupby('Month')
# Creates new dataframe "month" by grouping "df" by month
```

```python
## STEP 8: AGGREGATE
monthly_means = monthly['Daily_AirTemp_Mean_C'].mean()
monthly_means
```

```python
# Example: average a different column. Reads `monthly`; leaves monthly_means alone.
monthly_precip = monthly['Daily_Precip_Total_mm'].mean()
monthly_precip
```

```python
# Say it in a sentence!
july_temp = 11.2 # July's value, read from monthly_means
print(f"At Toolik, the average July temperature is {july_temp} degrees Celsius.")
```

```python
# Try it yourself--
bysite = df.groupby("LTER_Site")
wind_bysite = bysite['Daily_windsp_mean_msec'].mean()
wind_bysite
# There's one site
site = "ARC"
wind = "2.9955920734654424"
```

```python
print(f"At {site}, the average wind speed is {wind} msec.")
```

```python
# Try it yourself--
byyear = df.groupby("Year")
wind_byyear = byyear['Daily_windsp_mean_msec'].mean()
wind_byyear
```
