
## 2. The comparison operators


```python
print(temp > 0)
print(temp < 0)
print(temp >= 0)
print(temp <= 0)
print(temp == 8.4) # needs to use == when asking "is this equal to?"
# single "=" is an assginment
print(temp != 8.4) # bang operator, "is not"
```

```python
toolik.loc[0, 'Station'] # call a specific varible in row 0
```


## 3. Asking two questions at once


```python
# plain Python vernacular includes "and", "or", "not"
temp = 8.4
precip = 0.0

print(temp > 0 and precip == 0.0) # "and" returns true when BOTH sides are true
print(temp > 0 or precip > 0.0) # "or" returns true if ONE side is true
print(not temp > 0) # "not" returns opposite of true/false

# can also use $ and/or | in some Python packages
```


## 4. if / elif / else


```python
if temp > 0: # need a colon to trigger operation
    print("Below Freezing")
    # this indentation is how Python knows what's in a code block
```

```python
if temp > 0:
    print("Above Freezing")
else:
    print("Below Freezing")
```

```python
temp = -12.0
# elif combines else and if
# allows you to stack conditions and partition ranges

if temp > 10:
    category = 'hot'
elif temp > 5:
    category = 'mild'
elif temp > 0:
    category = 'cool'
elif temp > -5:
    category = 'cold'
else:
    category = 'brrrrr'
category
```


## 5. Asking a whole column


```python
for value in toolik.loc[[0 ,5000, 2], "Daily_AirTemp_Mean_C"]:
    if value > 7:
        print(f"{value} is warm")
    elif value > 5: 
        print(f"{value} is mild")
    else:
        print(f"{value} is brrrr")
    
```


## 6. Counting the Trues


```python
above_zero = toolik['Daily_AirTemp_Mean_C'] > 0
above_zero.sum()
```

```python
# pure Python versus methods
print(above_zero.sum() / toolik.shape[0]) # methods

print(above_zero.sum() / len(toolik))

print(sum(above_zero) / len(toolik)) # pure Python
```
