```python
import pandas as pd

url = 'https://eds-217-essential-python.github.io/data/openaq_goleta_measurments.csv'
goleta = pd.read_csv(url)

pm25 = goleta[goleta['parameter'] == 'pm25'].copy()
pm25.shape # pm25 only contains rows where the parameter is equal to 25
```

```python
from numpy import False_
pm25.sort_values('value').head() # default sort_values() starts with lowest value
pm25.sort_values('value', ascending = False_).head() # ascending = False means descending
```

```python
pm25.sort_values('datetimeLocal', ascending = False).head()
```

```python
# Sorting gives you the whole table in order. 
# Almost always you want the front of it, so pair the sort with .head():
pm25.sort_values('value', ascending=False).head(10)
# n = 10
```

```python
pm25.sort_values('value', ascending=True).head(5)
```

```python
goleta[goleta['parameter'] == 'pm10'].sort_values('value', ascending=False).head(5)[
    ['datetimeLocal', 'value', 'unit']
]
# Before you report a “top ten”, ask yourself what those ten rows are being compared against, 
# and whether every row in the table belongs in that comparison. 
# Yesterday’s units problem and today’s ranking problem come from the same mistake, 
# which is putting numbers that measure different things next to each other 
# and treating them as one set.
```

```python
# .idxmax() gives you the index label of the largest value in a column:
# .idxmin() does the same for the smallest value:

pm25['value'].idxmax()
pm25['value'].idxmin()
```

```python
# The number you get back is a row label, not a value and not a position. 
# Hand it to .loc[] with a column name and you get the one thing you were after:

worst_hour = pm25['value'].idxmax()
pm25.loc[worst_hour, 'datetimeLocal'] 
# .loc[row_label, column_label] reads as “the value at this row, in this column”. 
```

```python
cleanest_hour = pm25['value'].idxmin()
print(cleanest_hour)
print(pm25.loc[cleanest_hour, 'datetimeLocal'])
print(pm25.loc[cleanest_hour, 'value'])
```

```python
# Two hours in this dataset are tied at the maximum of 22 µg/m³, 
# and .idxmax() names only one of them. 
# Nothing in the output tells you that the second one is there.

# Whenever the answer to your question would change if there were a tie, 
# use the top-N pattern and read the result, rather than trusting a single label:

pm25.sort_values('value', ascending=False).head(3)[['datetimeLocal', 'value']]
```

## Putting it all together

```python
pm10 = goleta[goleta['parameter'] == 'pm10'].copy()
# Make a copy to ACTUALLY commit the dataframe to system memory
# Assignments only do not store dfs, only calls the original assignment code
# If you want to permanently manipulate a df into a new df for use, then .copy() it
worst_ten = pm10.sort_values('value', ascending=False).head(10)
worst_ten[['datetimeLocal', 'value', 'unit']]
```

```python
worst = pm10['value'].idxmax()
print('Worst pm10 hour:', pm10.loc[worst, 'datetimeLocal'])
print('Value:', pm10.loc[worst, 'value'], pm10.loc[worst, 'unit'])
```
